```python
#loading the dataset and removing additional info column

import pandas as pd
data=pd.read_csv('used_car_dataset.csv')
data.drop(columns=['AdditionInfo'],inplace=True)

data.columns
```

```python
data.head()
```

```python
data.info()
```

```python
data.drop(columns=['PostedDate'],inplace=True)
data.info()
```

```python
data['FuelType'].value_counts()
```

```python
data['Transmission'].value_counts()
```

```python
data['Owner'].value_counts()
```

```python
# since the above columns have only 2 or 3 types of values instead of using one hot encoding or binary classifier we will 
# manually classify them with values 0,1,2

data['Transmission']=data['Transmission'].map({'Manual':0,'Automatic':1})
data['Owner']=data['Owner'].map({'first':0,'second':1})
data['FuelType']=data['FuelType'].map({'Petrol':0,'Diesel':1,'Hybrid/CNG':2})

data.head(5)
```

```python
data.drop(columns=['PostedDate'],inplace=True)
data.head()
```

```python
# now we will convert the columns kmdriven and askprice to pure float/int columns
data['AskPrice']=data['AskPrice'].str.replace(r'[₹,\s]','',regex=True).astype(int)
data.head()
```

```python
cleaned_text=data['kmDriven'].str.replace(r'[^\d.]','',regex=True)
data['kmDriven']=pd.to_numeric(cleaned_text,errors='coerce')

data.head()
```

```python
data.info()
```

```python
data['Brand'].value_counts().count()
```

```python
# now we will create one column which will include both brand and model
data['brand_model']=data['Brand']+' '+data['model']
data.info()
```
